# 05 - Remaining Useful Life (RUL) Modeling (NASA C-MAPSS FD001)
Engine-wise regression modeling comparing Random Forest and XGBoost with NASA PHM scoring.

```python
import sys
import os
import joblib
import pandas as pd
import numpy as np

sys.path.insert(0, os.path.abspath('..'))
from src.rul_model import RULPredictionPipeline

pipe = RULPredictionPipeline(subset='FD001', max_rul=125, random_state=42)
results = pipe.train_and_evaluate()
```

## Model Comparison on Blind Test Engines (100 Turbofans)

```python
rows = []
for m in ['Random_Forest', 'XGBoost']:
    t = results[m]['test']
    v = results[m]['validation']
    rows.append({
        'Model': m,
        'Val RMSE': f"{v['rmse']:.2f}",
        'Test RMSE': f"{t['rmse']:.2f}",
        'Test MAE': f"{t['mae']:.2f}",
        'Test R2': f"{t['r2_score']:.3f}",
        'NASA Score': f"{t['nasa_total_score']:.1f}"
    })
pd.DataFrame(rows)
```
